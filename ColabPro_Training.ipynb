# Load Source




```python
from google.colab import drive
GoogleDriveRootDir = '/content/drive'                 # Google Drive的根目录
GoogleDriveDataDir = '{}/My Drive'.format(GoogleDriveRootDir)     # Google Drive的资源目录
drive.mount(GoogleDriveRootDir)                     # 载入云盘中的数据
import os
PythonFileDir = os.path.join(GoogleDriveDataDir,'DFDFNet')  # 主程序所在目录
PythonFileList = os.listdir(PythonFileDir) # 显示主程序目录内容
for file in PythonFileList:
  if file.endswith('.py'):
    print('Files：{}'.format(file))
  elif os.path.isdir(os.path.join(PythonFileDir,file)):
    print('SubDir：{} ({})'.format(file, os.listdir(os.path.join(PythonFileDir,file))))
os.chdir(PythonFileDir)  # 切除当前目录到主程序目录
```
