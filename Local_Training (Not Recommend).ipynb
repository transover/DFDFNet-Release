### **Example Dataset: Ninapro DB2**

This repository ships **DB2** as the reference example dataset. DB2 is the second sub-database of the Ninapro benchmark for surface electromyography (sEMG) gesture recognition. It contains 40 able-bodied subjects, each performing 49 hand/wrist gestures (Exercises A, B, and C) while 12 Delsys Trigno sEMG channels are recorded (2000 Hz). Every gesture is repeated 6 times.

The data is preprocessed into two branches consumed by DFDFNet:

- **Envelope branch** — the raw temporal signal after envelope rectification, shaped as `(samples, time, channels)`.
- **Feature branch** — a hand-crafted feature set, shaped as `(samples, features, channels)`.

Both branches follow the layout below, where `E1/E2/E3` denote the exercise and the trailing number is the repetition index:

```
Data/DB2/
├── DB2_Envelope/Results_200ms/Ninapro_Data_S1_A1_E123/data_npy.npy
└── DB2_Feature/Results_200ms/Ninapro_Data_S1_A1_E123/data_npy.npy
```

### One-click testing program

````python
# CLI Use: python train.py --database DB4 --exercise All --subject-list "[1]" --epochs 200
import subprocess, sys, os
env = os.environ.copy()
env['PYTHONUNBUFFERED'] = '1'

cmd = [
    sys.executable, "-u",
    "{}.py".format('launch'),                       # main program name
    # ---- Model params ----
    "--model-name={}".format('DFDFNet'),            # model name
    "--fusion-weight={}".format(0.5),               # fusion weight
    "--fusion-mode={}".format(1),                   # fusion mode 1/2/3
    "--lambda-s-trainable",                         # fusion weight trainable
    "--drb-fusion",                                 # use DRB clustering output
    "--ffn-fusion",                                 # use FFN output
    "--setcn-branch",                               # use SETCN feature branch
    "--irb-branch",                                 # use IRB envelope branch
    # ---- Data params ----
    "--database={}".format('DB2'),                  # dataset name 
    "--source-dir={}".format('./Data'),             # data source root directory
    "--window-length={}".format(200),               # window length (ms)
    "--exercise={}".format('All'),                  # gesture subset A/B/C/All
    "--subject-list={}".format('[1]'),              # subject list
    # "--channel-list={}".format('[]'),              # channel list
    "--process-method={}".format('None'),           # preprocessing method
    "--train-split-method={}".format('Repeat'),     # split method Random/Repeat/FoldK
    # "--test-split={}".format(0.2),                # validation ratio for random split
    # ---- Training params ----
    "--epochs={}".format(200),                      # number of epochs
    "--batch-size={}".format(320),                  # batch size
    # "--steps-per-epoch={}".format(None),          # steps per epoch
    # "--validation-steps={}".format(None),         # validation steps per epoch
    "--learning-rate={}".format(1e-3),              # initial learning rate
    "--loss={}".format('margin'),                   # loss function
    # "--reload-checkpoint",                        # whether to resume from checkpoint 
    "--verbose={}".format(1),                       # log verbosity level
    "--output-dir={}".format('models'),             # result output directory
    "--seed={}".format(42),                         # random seed
]

p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                    bufsize=1, text=True, env=env, encoding='utf8')

try:
    while True:
        line = p.stdout.readline()
        if not line and p.poll() is not None:
            break
        if line:
            print(line, end="")
            sys.stdout.flush()
except KeyboardInterrupt:
    p.terminate()
    raise
finally:
    remaining = p.stdout.read()
    if remaining:
        print(remaining, end="")
    p.wait() 
````
